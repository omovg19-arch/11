## Part 2. Descriptive statistics

Descriptive statistics summarise a variable using a small set of measures.

- **Centre:** mean, median, mode.
- **Spread:** standard deviation, variance, minimum, maximum, range and interquartile range (IQR).
- **Shape:** skewness, examined alongside a histogram or box plot.

`pandas` computes the **sample** standard deviation and sample variance by default (denominator `n - 1`); this differs from the population formula (denominator `n`).

## Quick reference: essential `pandas` and plotting commands

| Goal | Python command |
|:--|:--|
| Inspect data | `df.head()`, `df.info()` |
| Group means | `df.groupby('Category')['Sales'].mean()` |
| Summary statistics | `df[['Sales', 'Advertising_Spend']].describe()` |
| Mean / median / mode | `.mean()`, `.median()`, `.mode()` |
| Sample standard deviation / variance | `.std()`, `.var()` |
| Quartiles / IQR | `.quantile(0.25)`, `.quantile(0.75)` and subtract |
| Skewness | `.skew()` |
| Association | `.corr()` |
| Plots | `plt.plot()`, `plt.bar()`, `plt.scatter()`, `plt.hist()`, `plt.boxplot()` |
